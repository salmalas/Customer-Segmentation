## **Clustering via KMeans** ##

```python
features = ["Age", "Income","Total_Spendings","NumWebPurchases" ,"NumStorePurchases", "NumWebVisitsMonth","Recency"]
```

```python
X=df[features].copy()
X
```

preprocessing: scaling



```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
from sklearn.cluster import KMeans
list = []

for i in range(2,10):
    kmeans = KMeans(n_clusters = i)
    kmeans.fit(X_scaled)
    list.append(kmeans.inertia_)

list
```

```python
plt.plot(range(2,10),list, marker = "o")
plt.ylabel("NUM of Clusters ")
plt.xlabel("inertia")
plt.title("Elbow metthod for optimal k")
```

```python
kmeans = KMeans(n_clusters=5)
df["Cluster"]= kmeans.fit_predict(X_scaled)
```

```python
df["Cluster"]
```

```python
df.head()
```

```python
cluster_analysis= df.groupby("Cluster")[features].mean()
cluster_analysis
```

```python
df["Cluster"].value_counts()
```

Using PCA inorder to reduce feature dimensions to 2
: all rows , 0 1 forst or second col


```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
PCADATA = pca.fit_transform(X_scaled)
df["PCA1"]= PCADATA[:,0]
df["PCA2"]= PCADATA[:,1]

```

```python
sns.scatterplot(x="PCA1",y="PCA2",hue="Cluster",data=df,palette="Set1")
plt.title("Customer Segmenation PCA")
```

```python
cluster_names = {
    0: "Low Value Customers",
    1: "Young High Value Customers",
    2: "Budget Customers",
    3: "Frequent Buyers",
    4: "Old High Value Customers"
}
df["Customer_Segment"] = df["Cluster"].map(cluster_names)
```
